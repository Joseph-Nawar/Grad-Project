# 6.6 Acute Symptom Assessment Design

Purpose:
Document and test the FAST/BE-FAST-inspired acute symptom module with a synthetic scenario audit.

This module is deterministic, not trained ML, because no suitable public paired acute stroke questionnaire dataset is available.

The notebook now complements the reusable scenario audit script instead of regenerating the design report inline.

```python
from pathlib import Path
import subprocess
import sys

import pandas as pd

sys.path.insert(0, str(Path.cwd().parent))

from rural_stroke_assist.config import ACUTE_SYMPTOM_ASSESSMENT_REPORT_PATH

from rural_stroke_assist.capture.acute_symptom_schema import AcuteStrokeSymptoms
from rural_stroke_assist.modules.acute_symptom_module import (
    assess_acute_stroke_symptoms,
)
```

```python
low_case = AcuteStrokeSymptoms()

low_result = assess_acute_stroke_symptoms(low_case)

low_result
```

```python
moderate_case = AcuteStrokeSymptoms(
    vision_disturbance=True,
    balance_or_coordination_loss=True,
)

moderate_result = assess_acute_stroke_symptoms(moderate_case)

moderate_result
```

```python
urgent_case = AcuteStrokeSymptoms(
    face_drooping=True,
    arm_weakness=True,
    speech_difficulty=True,
    symptom_onset_minutes=60,
)

urgent_result = assess_acute_stroke_symptoms(urgent_case)

urgent_result
```

```python
audit_script = Path.cwd().parent / "scripts" / "acute_symptom_scenario_audit.py"

subprocess.run([sys.executable, str(audit_script)], check=True)

audit_path = ACUTE_SYMPTOM_ASSESSMENT_REPORT_PATH.parent / "acute_symptom_scenario_audit.csv"
audit_df = pd.read_csv(audit_path)

audit_df[[
    "scenario_name",
    "acute_symptom_score",
    "risk_band",
    "expected_min_band",
    "passes_expectation",
]].head(12)
```
