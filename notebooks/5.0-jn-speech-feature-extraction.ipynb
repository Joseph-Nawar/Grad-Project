# 5.0 Speech Feature Extraction

Purpose:
- Load speaker-safe speech split manifest
- Extract MFCC and basic acoustic features
- Save modeling-ready speech feature table

```python
import pandas as pd
from tqdm import tqdm

from pathlib import Path
import sys

sys.path.insert(0, str(Path.cwd().parent))

tqdm.pandas()

from rural_stroke_assist.config import (
    SPEECH_SPLIT_MANIFEST_PATH,
    SPEECH_FEATURES_PATH,
    SPEECH_FEATURE_EXTRACTION_FAILURES_PATH
)
from rural_stroke_assist.features.speech_features import (
    build_speech_features_table,
)
```

```python
speech_split_df = pd.read_csv(SPEECH_SPLIT_MANIFEST_PATH)

speech_split_df.head()
```

```python
speech_split_df.groupby(["split", "label"]).size()
```

```python
speech_features_df, failed_features_df = build_speech_features_table(
    split_manifest=speech_split_df,
    target_sample_rate=16000,
    max_duration_seconds=5.0,
)

speech_features_df.head()
```

```python
speech_features_df.shape
```

```python
speech_features_df.isna().sum().sort_values(ascending=False).head(20)
```

```python
SPEECH_FEATURES_PATH.parent.mkdir(parents=True, exist_ok=True)

speech_features_df.to_csv(SPEECH_FEATURES_PATH, index=False)
failed_features_df.to_csv(SPEECH_FEATURE_EXTRACTION_FAILURES_PATH, index=False)

print("Saved features to:", SPEECH_FEATURES_PATH)
print("Saved failures to:", SPEECH_FEATURE_EXTRACTION_FAILURES_PATH)
```
