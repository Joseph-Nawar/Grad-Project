```python
from collections import Counter

face_file_extensions = Counter(
    path.suffix.lower()
    for path in FACE_DATA_DIR.rglob("*")
    if path.is_file()
)

metadata_file_extensions = Counter(
    path.suffix.lower()
    for path in METADATA_DATA_DIR.rglob("*")
    if path.is_file()
)

print("Face dataset file extensions:", face_file_extensions)
print("Metadata dataset file extensions:", metadata_file_extensions)
```
